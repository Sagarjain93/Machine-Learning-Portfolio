**Check for outliers**

**Users — Age**

````python
# Function to calculate IQR-based outliers

def outlier_summary(df, column):
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = ((df[column] < lower_bound) |
                (df[column] > upper_bound)).sum()

    return Q1, Q3, IQR, lower_bound, upper_bound, outliers


Q1, Q3, IQR, lower, upper, outliers = outlier_summary(users, 'age')

print("Users - Age")
print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Bound:", lower)
print("Upper Bound:", upper)
print("Potential Outliers:", outliers)
````

**Flights**

````python
# Check potential outliers in flight numerical variables

for col in ['price', 'time', 'distance']:
    Q1, Q3, IQR, lower, upper, outliers = outlier_summary(flights, col)

    print(f"\nFlights - {col}")
    print("Q1:", Q1)
    print("Q3:", Q3)
    print("IQR:", IQR)
    print("Lower Bound:", lower)
    print("Upper Bound:", upper)
    print("Potential Outliers:", outliers)
````

**Hotels**

````python
# Check potential outliers in hotel numerical variables

for col in ['days', 'price', 'total']:
    Q1, Q3, IQR, lower, upper, outliers = outlier_summary(hotels, col)

    print(f"\nHotels - {col}")
    print("Q1:", Q1)
    print("Q3:", Q3)
    print("IQR:", IQR)
    print("Lower Bound:", lower)
    print("Upper Bound:", upper)
    print("Potential Outliers:", outliers)
````

#### **Chart 17 — Average Flight Price by Top Routes**

##### **1. Why did you pick the specific chart?**

A horizontal bar chart makes the longer route names easier to read and compare.

The average flight price varies considerably across the top 10 high-volume routes.

Sao Paulo (SP) → Florianopolis (SC) has the highest average flight price, at approximately 1,380, followed by Brasilia (DF) → Florianopolis (SC) at approximately 1,260. Florianopolis (SC) → Aracaju (SE) and Aracaju (SE) → Florianopolis (SC) also have relatively high average prices of approximately 1,250 and 1,230 respectively.

## ***5. Hypothesis Testing***

### Based on your chart experiments, define three hypothetical statements from the dataset. In the next three questions, perform hypothesis testing to obtain final conclusion about the statements through your code and statistical testing.

Answer Here.

###  **Hypothetical Statement - 1**

**Does flight type have a statistically significant effect on flight price?**

#### 1. State Your research hypothesis as a null hypothesis and alternate hypothesis.

**Null Hypothesis (H₀):**
There is no statistically significant difference in the mean flight prices among economic, premium, and first-class flights.

**Alternative Hypothesis (H₁):**
There is a statistically significant difference in the mean flight price among at least one of the flight types.

**Significance Level (α):** 0.05

#### 2. Perform an appropriate statistical test.

````python
# Hypothesis Testing - 1
# One-way ANOVA: Flight Type vs Flight Price

from scipy.stats import f_oneway

# Create price groups for each flight type
economic_price = flights.loc[
    flights['flightType'] == 'economic',
    'price'
]

premium_price = flights.loc[
    flights['flightType'] == 'premium',
    'price'
]

first_class_price = flights.loc[
    flights['flightType'] == 'firstClass',
    'price'
]

# Perform one-way ANOVA
f_statistic, p_value = f_oneway(
    economic_price,
    premium_price,
    first_class_price
)

print("One-way ANOVA Test")
print("-" * 30)
print("F-statistic:", round(f_statistic, 4))
print("P-value:", p_value)

# Decision
alpha = 0.05

if p_value < alpha:
    print("\nDecision: Reject the Null Hypothesis (H₀).")
    print("Conclusion: Flight type has a statistically significant")
    print("difference in mean flight prices.")
else:
    print("\nDecision: Fail to Reject the Null Hypothesis (H₀).")
    print("Conclusion: There is insufficient evidence to conclude")
    print("that mean flight prices differ by flight type.")
````

##### Which statistical test have you done to obtain P-Value?

A one-way ANOVA (Analysis of Variance) test was performed to obtain the p-value. The test was used to determine whether there is a statistically significant difference in the mean flight prices among economic, premium, and first-class flights.

##### Why did you choose the specific statistical test?

A one-way ANOVA test was chosen because the analysis compares the mean of a numerical variable, flight price, across three independent categorical groups: economic, premium, and first-class flights.

ANOVA is appropriate for testing whether there is a statistically significant difference among the means of more than two groups. Therefore, it is more suitable than performing multiple pairwise t-tests.

### Final Conclusion

The p-value obtained from the one-way ANOVA test is effectively 0, which is less than the significance level of 0.05. Therefore, we reject the null hypothesis (H₀).

There is statistically significant evidence that mean flight prices differ among the economic, premium, and first-class flight types.

This statistical result supports the earlier visualization findings, where first-class flights showed the highest average prices, followed by premium and economic flights.

### 2. Handling Outliers

````python
# Handling Outliers & Outlier Treatments

# Checking potential outliers using the IQR method

def check_outliers_iqr(df, columns):
    outlier_summary = {}

    for column in columns:
        Q1 = df[column].quantile(0.25)
        Q3 = df[column].quantile(0.75)
        IQR = Q3 - Q1

        lower_bound = Q1 - 1.5 * IQR
        upper_bound = Q3 + 1.5 * IQR

        outliers = df[(df[column] < lower_bound) | (df[column] > upper_bound)]

        outlier_summary[column] = len(outliers)

    return outlier_summary


# Numerical columns for outlier checking
users_numeric = ['age']

flights_numeric = ['price', 'time', 'distance']

hotels_numeric = ['days', 'price', 'total']


print("Potential outliers in Users dataset:")
print(check_outliers_iqr(users, users_numeric))

print("\nPotential outliers in Flights dataset:")
print(check_outliers_iqr(flights, flights_numeric))

print("\nPotential outliers in Hotels dataset:")
print(check_outliers_iqr(hotels, hotels_numeric))
````

##### What all outlier treatment techniques have you used and why did you use those techniques?

Potential outliers were checked using the Interquartile Range (IQR) method for the numerical variables in the Users, Flights, and Hotels datasets. The analysis identified zero potential outliers in all checked variables.

## ***7. ML Model Implementation***

### ML Model - 1

Import Libraries

````python
# Import required libraries

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from scipy.stats import f_oneway, pearsonr

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

import warnings
warnings.filterwarnings('ignore')
````

````python
# Prepare data for Flight Price Regression

# Create a copy so that the original cleaned dataset remains unchanged
regression_data = flights.copy()

# Extract useful features from the date column
regression_data['year'] = regression_data['date'].dt.year
regression_data['month'] = regression_data['date'].dt.month
regression_data['dayofweek'] = regression_data['date'].dt.dayofweek
regression_data['quarter'] = regression_data['date'].dt.quarter

# Define target variable
y = regression_data['price']

# Define predictor variables
X = regression_data[
    [
        'from',
        'to',
        'flightType',
        'agency',
        'time',
        'distance',
        'year',
        'month',
        'dayofweek',
        'quarter'
    ]
].copy()

print("Feature dataset shape:", X.shape)
print("Target dataset shape:", y.shape)

print("\nFeatures:")
print(X.columns.tolist())

print("\nTarget variable:")
print(y.name)
````

````python
# Split the data into training and testing sets

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("Training feature shape:", X_train.shape)
print("Testing feature shape:", X_test.shape)

print("Training target shape:", y_train.shape)
print("Testing target shape:", y_test.shape)
````

````python
# Define categorical and numerical features

categorical_features = [
    'from',
    'to',
    'flightType',
    'agency'
]

numerical_features = [
    'time',
    'distance',
    'year',
    'month',
    'dayofweek',
    'quarter'
]


# Create preprocessing pipeline

preprocessor = ColumnTransformer(
    transformers=[
        (
            'categorical',
            OneHotEncoder(handle_unknown='ignore'),
            categorical_features
        ),
        (
            'numerical',
            StandardScaler(),
            numerical_features
        )
    ]
)

print("Categorical features:", categorical_features)
print("Numerical features:", numerical_features)
print("\nPreprocessing pipeline created successfully.")
````

````python
from sklearn.linear_model import LinearRegression
````

````python
# ML Model 1 - Linear Regression

# Create the Linear Regression pipeline
linear_regression_pipeline = Pipeline(
    steps=[
        ('preprocessor', preprocessor),
        ('model', LinearRegression())
    ]
)

# Fit the model on training data
linear_regression_pipeline.fit(X_train, y_train)

print("Linear Regression model trained successfully.")
````

````python
# Predict on the test dataset

y_pred_lr = linear_regression_pipeline.predict(X_test)

print("Predictions generated successfully.")
print("First 10 predictions:")
print(y_pred_lr[:10])
````

````python
# 5-Fold Cross-Validation for Linear Regression

cv_results_lr = cross_validate(
    linear_regression_pipeline,
    X_train,
    y_train,
    cv=5,
    scoring={
        'MAE': 'neg_mean_absolute_error',
        'RMSE': 'neg_root_mean_squared_error',
        'R2': 'r2'
    },
    n_jobs=-1
)

# Convert negative error scores to positive values
cv_mae_lr = -cv_results_lr['test_MAE']
cv_rmse_lr = -cv_results_lr['test_RMSE']
cv_r2_lr = cv_results_lr['test_R2']

print("Linear Regression - 5-Fold Cross-Validation")
print("-" * 50)

print("Mean MAE :", round(cv_mae_lr.mean(), 4))
print("Mean RMSE:", round(cv_rmse_lr.mean(), 4))
print("Mean R²  :", round(cv_r2_lr.mean(), 4))

print("\nStandard Deviation")
print("MAE :", round(cv_mae_lr.std(), 4))
print("RMSE:", round(cv_rmse_lr.std(), 4))
print("R²  :", round(cv_r2_lr.std(), 4))
````

Hyperparameter optimization will be applied to the more complex regression models evaluated later in the project, where tuning can meaningfully improve model performance.

The R² score increased from 0.9193 to 0.9194, while both MAE and RMSE decreased slightly. The low standard deviation across the folds also indicates that the model performance is stable and consistent.

Since Linear Regression has no significant hyperparameters requiring optimization, no hyperparameter tuning was performed for this baseline model. More extensive hyperparameter optimization will be performed on the subsequent machine learning models.
